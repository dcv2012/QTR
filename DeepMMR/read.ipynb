# 使用说明
## 1. 基本结构
- configs: 存储实验参数配置文件，用于设置模型超参数（如学习率、批大小、网络深度等）。这些JSON文件在实验运行时被加载，以配置不同模型和场景。 
- data: 存储数据集和数据预处理代码。包含真实数据集（如RHC）和模拟数据生成逻辑，用于训练和测试模型。
- src: 核心源代码文件夹，包含模型实现、数据处理、工具函数和实验逻辑。分为子文件夹以模块化组织。
    - src/data: 定义数据类和生成函数
    - src/models: 存放各种实现的模型
    - src/utils: 工具函数等
